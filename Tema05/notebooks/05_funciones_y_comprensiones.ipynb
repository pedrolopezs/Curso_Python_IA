## Funciones, return y varios valores

Compara una función que devuelve un valor con otra que solo lo muestra, devuelve dos valores y usa varios `return`.

```python
# Funciones, return y varios valores (Tema 5).
# Devuelve el resultado, por lo que se puede guardar y comparar
def convertir_a_gb(megabytes):
    return megabytes / 1024


# Solo muestra el resultado y no devuelve nada
def mostrar_en_gb(megabytes):
    print(megabytes / 1024)


# Devuelve dos valores, que Python agrupa en una tupla
def separar_direccion(direccion):
    ip, puerto = direccion.split(":")
    return ip, int(puerto)


# Varios return: se ejecuta el primero que se cumple
def clasificar_puerto(puerto):
    if puerto < 1024:
        return "sistema"
    if puerto < 49152:
        return "registrado"
    return "dinámico"


tamanio = convertir_a_gb(2048)
print(tamanio)
print(tamanio > 1)

resultado = mostrar_en_gb(2048)
print(resultado)

print(separar_direccion("10.0.0.5:443"))
ip, puerto = separar_direccion("10.0.0.5:443")
print(ip, puerto)

print(clasificar_puerto(22), clasificar_puerto(8080))
print(clasificar_puerto(50000))
```

## Argumentos de las funciones

Llama a una función con argumentos posicionales, por nombre y con valores por defecto, y prueba `*args`, `**kwargs` y `lambda`.

```python
# Argumentos de las funciones y expresiones lambda (Tema 5).
def crear_url(servidor, puerto=80, protocolo="http"):
    return f"{protocolo}://{servidor}:{puerto}"


# Posicionales, valores por defecto y argumentos por nombre
print(crear_url("web01"))
print(crear_url("web01", 8080))
print(crear_url("web01", protocolo="https", puerto=443))


# *usuarios recoge los argumentos posicionales sobrantes en una tupla
def contar_conectados(*usuarios):
    return len(usuarios)


print(contar_conectados("ana", "luis", "marta"))
print(contar_conectados())


# **datos recoge los argumentos con nombre sobrantes en un diccionario
def mostrar_ficha(equipo, **datos):
    print(equipo)
    for clave, valor in datos.items():
        print(f"  {clave}: {valor}")


mostrar_ficha("SW01", ubicacion="Sala 2", puertos=24)

# Una lambda como clave de ordenación
conexiones = [("web01", 120), ("db01", 45), ("web02", 300)]
print(sorted(conexiones, key=lambda conexion: conexion[1]))
print(max(conexiones, key=lambda conexion: conexion[1]))
```

## Docstrings y anotaciones de tipo

Consulta los docstrings con `__doc__` y observa las anotaciones de tipo de una función.

```python
# Docstrings, anotaciones de tipo y consulta de la documentación (Tema 5).
def calcular_media(valores):
    """Devuelve la media de una lista de números."""
    return sum(valores) / len(valores)


def calcular_ocupacion(usado: int, total: int, decimales: int = 1) -> float:
    """Calcula el porcentaje de disco ocupado.

    Parámetros:
        usado: gigabytes ocupados.
        total: capacidad del disco en gigabytes.
        decimales: decimales del resultado.

    Devuelve el porcentaje como número decimal.
    """
    return round(usado / total * 100, decimales)


print(calcular_media.__doc__)
print(calcular_media.__name__)
print(calcular_ocupacion.__doc__)
print(calcular_ocupacion(87, 250))
```

## Ámbito de las variables

Comprueba la diferencia entre una variable local y una global, y las dos formas de modificar un contador.

```python
# Variables locales y globales, global y su alternativa (Tema 5).
limite = 100
conexiones_activas = 0


# Asignar dentro de la función crea una variable local
def comprobar(valor):
    limite = 50
    return valor > limite


print(comprobar(70), limite)


# global hace que la función use la variable global
def registrar_conexion():
    global conexiones_activas
    conexiones_activas += 1


registrar_conexion()
registrar_conexion()
print(conexiones_activas)


# Alternativa recomendada: recibir el valor y devolver el resultado
def sumar_conexion(conexiones):
    return conexiones + 1


total = 0
total = sumar_conexion(total)
total = sumar_conexion(total)
print(total)
```

## Listas como argumentos y valor por defecto mutable

Observa cuándo una función modifica la lista que recibe y qué ocurre con una lista como valor por defecto.

```python
# Listas como argumentos y valor por defecto mutable (Tema 5).
# La función modifica la lista recibida
def agregar_regla(reglas, nueva):
    reglas.append(nueva)


# La función trabaja con una copia y devuelve la lista nueva
def ampliar_reglas(reglas, nueva):
    copia = reglas.copy()
    copia.append(nueva)
    return copia


firewall = ["permitir 22", "permitir 80"]
agregar_regla(firewall, "permitir 443")
print(firewall)
ampliadas = ampliar_reglas(firewall, "denegar 23")
print(ampliadas)
print(firewall)


# Error clásico: la lista por defecto se crea una sola vez
def anotar_alerta(mensaje, historial=[]):
    historial.append(mensaje)
    return historial


print(anotar_alerta("disco lleno"))
print(anotar_alerta("cpu alta"))


# Solución habitual: None como valor por defecto
def anotar_alerta_segura(mensaje, historial=None):
    if historial is None:
        historial = []
    historial.append(mensaje)
    return historial


print(anotar_alerta_segura("disco lleno"))
print(anotar_alerta_segura("cpu alta"))
```

## Comprensiones

Crea listas, un diccionario y un conjunto con comprensiones, y usa una expresión generadora.

```python
# Comprensiones de listas, diccionarios, conjuntos y generadoras (Tema 5).
latencias = [23, 145, 12, 230, 67]

# Lista con transformación
segundos = [latencia / 1000 for latencia in latencias]
print(segundos)

# Lista con filtro
lentas = [latencia for latencia in latencias if latencia > 100]
print(lentas)

# Expresión condicional: un valor distinto según la condición
estados = ["lenta" if latencia > 100 else "normal" for latencia in latencias]
print(estados)

# Diccionario a partir de un diccionario, con filtro
servicios = {"ssh": 22, "http": 80, "mysql": 3306}
reservados = {nombre: puerto for nombre, puerto in servicios.items()
              if puerto < 1024}
print(reservados)

# Conjunto: los duplicados desaparecen
direcciones = ["10.0.0.5", "10.0.0.7", "10.0.0.5"]
unicas = {direccion for direccion in direcciones}
print(sorted(unicas))

# Expresión generadora: un único resultado sin crear una lista intermedia
total_ms = sum(latencia for latencia in latencias)
print(total_ms)

# Una expresión generadora se recorre una sola vez
rapidas = (latencia for latencia in latencias if latencia < 50)
print(list(rapidas))
print(list(rapidas))
```

## Informe de un registro de accesos

Reúne funciones, una constante, docstrings y comprensiones en un script completo.

```python
# Informe de peticiones por dirección IP de un registro (Tema 5).
UMBRAL_PETICIONES = 3

registro = [
    "10.0.0.5 GET /index",
    "10.0.0.7 GET /admin",
    "10.0.0.5 POST /login",
    "10.0.0.9 GET /index",
    "10.0.0.5 GET /admin",
    "10.0.0.7 GET /index",
]


def extraer_direccion(linea):
    """Devuelve la dirección IP de una línea del registro."""
    return linea.split()[0]


def contar_peticiones(lineas):
    """Devuelve un diccionario con las peticiones de cada dirección."""
    totales = {}
    for linea in lineas:
        direccion = extraer_direccion(linea)
        totales[direccion] = totales.get(direccion, 0) + 1
    return totales


def clasificar(peticiones, umbral=UMBRAL_PETICIONES):
    """Devuelve 'sospechosa' si se alcanza el umbral y 'normal' si no."""
    if peticiones >= umbral:
        return "sospechosa"
    return "normal"


totales = contar_peticiones(registro)
for direccion in sorted(totales):
    peticiones = totales[direccion]
    print(f"{direccion:<9} {peticiones} {clasificar(peticiones)}")

sospechosas = [direccion for direccion, peticiones in totales.items()
               if clasificar(peticiones) == "sospechosa"]
print("Sospechosas:", sospechosas)

paginas = {linea.split()[2] for linea in registro}
print("Páginas:", sorted(paginas))
```

## Registro de eventos

Esta función tiene un valor por defecto mutable. Las dos llamadas deberían devolver listas distintas.
