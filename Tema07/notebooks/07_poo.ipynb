## Métodos y estado

Los métodos `describir()` y `apagar()` leen y modifican el estado con `self`.

```python
"""Métodos, self y cambio de estado (Tema 7)."""


class Equipo:
    def __init__(self, nombre, direccion_ip, estado="activo"):
        self.nombre = nombre
        self.direccion_ip = direccion_ip
        self.estado = estado

    def describir(self):
        return (f"{self.nombre} ({self.direccion_ip}): "
                f"{self.estado}")

    def apagar(self):
        self.estado = "apagado"


router = Equipo("router01", "10.0.0.1")
print(router.describir())
router.apagar()
print(router.describir())
```

## Representación y comparación

Sin `__str__()` ni `__eq__()`, `print()` muestra la dirección del objeto y `==` da `False`.

```python
"""Representación y comparación de objetos (Tema 7)."""


class Equipo:
    def __init__(self, nombre, direccion_ip):
        self.nombre = nombre
        self.direccion_ip = direccion_ip


router = Equipo("router01", "10.0.0.1")
copia = Equipo("router01", "10.0.0.1")
otro = Equipo("switch01", "10.0.0.2")
print(router)
print(router == copia, router == otro)
```

## Atributos de clase y lista compartida

Un atributo de clase lo comparten todas las instancias. Asignar a través de una instancia crea un atributo propio, mientras que `append()` modifica la lista compartida.

```python
"""Atributos de clase, atributos de instancia y lista compartida."""


class Equipo:
    total = 0

    def __init__(self, nombre):
        self.nombre = nombre
        Equipo.total += 1


router = Equipo("router01")
switch = Equipo("switch01")
print(Equipo.total, router.total)
switch.total = 10
print(Equipo.total, switch.total)


class Servidor:
    servicios = []

    def __init__(self, nombre):
        self.nombre = nombre

    def agregar_servicio(self, servicio):
        self.servicios.append(servicio)


web = Servidor("web01")
db = Servidor("db01")
web.agregar_servicio("ssh")
print(db.servicios)
```

## Encapsulación en Python

Un guion bajo avisa de que un atributo es interno, pero Python no impide modificarlo. Con dos guiones bajos, Python cambia el nombre del atributo dentro de la clase, aunque tampoco impide el acceso.

```python
"""Encapsulación con uno y con dos guiones bajos (Tema 7)."""


class Equipo:
    def __init__(self, nombre):
        self.nombre = nombre
        self._estado = "activo"
        self.__clave = "abc123"

    def cambiar_estado(self, estado):
        if estado not in ("activo", "apagado"):
            raise ValueError(f"Estado no válido: {estado}")
        self._estado = estado

    def describir(self):
        return f"{self.nombre}: {self._estado}"

    def comprobar_clave(self, clave):
        return clave == self.__clave


router = Equipo("router01")
router.cambiar_estado("apagado")
print(router.describir())
try:
    router.cambiar_estado("roto")
except ValueError as error:
    print(error)
router._estado = "roto"
print(router.describir())
print(router.comprobar_clave("abc123"))
try:
    print(router.__clave)
except AttributeError as error:
    print(error)
print(router._Equipo__clave)
```

## Composición

Un `Inventario` contiene una lista de objetos `Equipo`.

```python
"""Composición: un inventario formado por equipos (Tema 7)."""


class Equipo:
    def __init__(self, nombre, direccion_ip):
        self.nombre = nombre
        self.direccion_ip = direccion_ip


class Inventario:
    def __init__(self):
        self.equipos = []

    def agregar(self, equipo):
        self.equipos.append(equipo)

    def buscar(self, nombre):
        for equipo in self.equipos:
            if equipo.nombre == nombre:
                return equipo
        return None


inventario = Inventario()
inventario.agregar(Equipo("router01", "10.0.0.1"))
inventario.agregar(Equipo("switch01", "10.0.0.2"))
print(inventario.buscar("switch01").direccion_ip)
print(inventario.buscar("nas01"))
```

## Herencia y polimorfismo

`Servidor` hereda de `Equipo`, llama a `super().__init__()` y sobrescribe `describir()`.

```python
"""Herencia, super() y polimorfismo (Tema 7)."""


class Equipo:
    def __init__(self, nombre, direccion_ip):
        self.nombre = nombre
        self.direccion_ip = direccion_ip

    def describir(self):
        return f"{self.nombre} ({self.direccion_ip})"


class Servidor(Equipo):
    def __init__(self, nombre, direccion_ip):
        super().__init__(nombre, direccion_ip)
        self.servicios = ["ssh"]

    def describir(self):
        return (f"{super().describir()}, "
                f"servicios: {len(self.servicios)}")


equipos = [Equipo("router01", "10.0.0.1"),
           Servidor("srv01", "10.0.0.10")]
for equipo in equipos:
    print(equipo.describir())
print(isinstance(equipos[1], Equipo), isinstance(equipos[0], Servidor))
```

## Excepción propia

`IncidenciaError` hereda de `Exception` y se lanza con `raise`.

```python
"""Excepción propia IncidenciaError (Tema 7)."""


class IncidenciaError(Exception):
    """Error al crear una incidencia."""


def abrir_incidencia(prioridad):
    if prioridad not in ("alta", "media", "baja"):
        raise IncidenciaError(
            f"Prioridad no válida: {prioridad}")
    return prioridad


try:
    abrir_incidencia("urgente")
except IncidenciaError as error:
    print(error)

print(issubclass(IncidenciaError, Exception))

try:
    abrir_incidencia("urgente")
except Exception as error:
    print(type(error).__name__)
```

## Informe de incidencias

El script integrador lee `Tema07/datos/incidencias.txt` y cuenta las incidencias aceptadas y las descartadas.

```python
"""Informe de incidencias con clases y excepción propia (Tema 7)."""


class IncidenciaError(Exception):
    """Error al crear una incidencia."""


class Incidencia:
    PRIORIDADES = ("critica", "alta", "media", "baja")

    def __init__(self, codigo, equipo, prioridad, horas):
        if prioridad not in self.PRIORIDADES:
            raise IncidenciaError(f"Prioridad no válida: {prioridad}")
        self.codigo = codigo
        self.equipo = equipo
        self.prioridad = prioridad
        self.horas = horas


class Registro:
    def __init__(self):
        self.incidencias = []
        self.descartadas = 0

    def agregar_linea(self, linea):
        try:
            codigo, equipo, prioridad, horas = linea.split()
            incidencia = Incidencia(codigo, equipo, prioridad, int(horas))
        except (ValueError, IncidenciaError):
            self.descartadas += 1
        else:
            self.incidencias.append(incidencia)


registro = Registro()
with open("Tema07/datos/incidencias.txt", encoding="utf-8") as fichero:
    for linea in fichero:
        registro.agregar_linea(linea)
print(len(registro.incidencias), registro.descartadas)
```
