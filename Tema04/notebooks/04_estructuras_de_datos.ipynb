## Índices y slicing de cadenas

Accede a los caracteres con índices positivos y negativos, extrae fragmentos con slicing y prueba el salto y la inversión. Cambia el valor de `codigo` y comprueba cómo cambia cada resultado.

```python
# Índices y slicing de cadenas (Tema 4).
codigo = "SRV-01-A"

# Índices positivos y negativos
print(codigo[0])
print(codigo[-1])
print(len(codigo))

# Slicing [inicio:fin:salto]
print(codigo[0:3])
print(codigo[4:6])
print(codigo[:3])
print(codigo[4:])
print(codigo[-1:])

# Salto e inversión
print(codigo[::2])
print(codigo[::-1])

# Una cadena nueva a partir de fragmentos de la original
nuevo = "XRV" + codigo[3:]
print(nuevo)
```

## Listas

Agrega, elimina, busca y ordena ítems, y comprueba qué métodos modifican la lista y cuáles devuelven una lista nueva.

```python
# Listas: acceso, métodos y orden (Tema 4).
equipos = ["Portátil", "Switch", "Router"]

# Acceso y slicing
print(equipos[0], equipos[-1])
print(equipos[1:])

# Agregar ítems
equipos.append("NAS")
equipos.insert(1, "Servidor")
equipos.extend(["SAI", "Firewall"])
print(equipos)

# Eliminar y buscar
equipos.remove("Switch")
ultimo = equipos.pop()
print(ultimo, equipos)
print(equipos.index("Router"), "NAS" in equipos)

# sorted() devuelve una lista nueva y sort() modifica la original
print(sorted(equipos))
print(equipos)
equipos.sort()
print(equipos)

# Los métodos que modifican la lista devuelven None
resultado = equipos.append("Hub")
print(resultado)

# Recorrido con enumerate
for pos, equipo in enumerate(equipos, start=1):
    print(pos, equipo)
```

## Igualdad, identidad, alias y copias

Compara listas con `==` e `is` y observa el efecto de un alias, de una copia superficial y de una copia profunda.

```python
# Igualdad, identidad, alias y copias (Tema 4).
import copy

a = [1, 2, 3]
b = [1, 2, 3]
print(a == b, a is b)

# Alias: dos nombres para la misma lista
c = a
c.append(4)
print(a, c, a is c)

# Copia superficial
d = a.copy()
d.append(5)
print(a, d, a is d)

# La copia superficial comparte los objetos internos
inventario = [["PC", 12], ["Switch", 7]]
copia = inventario.copy()
copia[0][1] = 0
print(inventario)

# Copia profunda
profunda = copy.deepcopy(inventario)
profunda[0][1] = 99
print(inventario, profunda)
```

## Tuplas

Accede a una tupla, desempaquétala e intercambia dos valores.

```python
# Tuplas y desempaquetado (Tema 4).
lugar = ("Sala 2", "Armario B", 4)
print(lugar[0], lugar[-1], len(lugar))
print(lugar[:2])

# Desempaquetado
sala, armario, balda = lugar
print(f"{sala} / {armario} / balda {balda}")
primero, *resto = [12, 7, 25]
print(primero, resto)

# Intercambio de valores
a, b = 1, 2
a, b = b, a
print(a, b)

# Tupla de un solo ítem
un_item = ("SRV01",)
print(type(un_item), type(("SRV01")))

# enumerate() entrega tuplas
for par in enumerate(["PC", "AP"], start=1):
    print(par)
```

## Conjuntos

Elimina duplicados con un conjunto y aplica operaciones entre conjuntos.

```python
# Conjuntos: unicidad y operaciones (Tema 4).
marcas = ["HP", "Dell", "HP", "Lenovo", "Dell"]
unicas = set(marcas)
print(len(marcas), len(unicas))
print(sorted(unicas))

unicas.add("Acer")
unicas.discard("Asus")
print(sorted(unicas))
print("HP" in unicas)

# Operaciones entre conjuntos
almacen_a = {"SRV01", "SRV02", "SW01"}
almacen_b = {"SRV02", "SW01", "AP01"}
print(sorted(almacen_a | almacen_b))
print(sorted(almacen_a & almacen_b))
print(sorted(almacen_a - almacen_b))
print(sorted(almacen_a ^ almacen_b))
print({"SRV01"} <= almacen_a)

# {} crea un diccionario vacío
print(type({}), type(set()))
```

## Diccionarios

Consulta, modifica y recorre un diccionario, y cuenta ocurrencias con `get()`.

```python
# Diccionarios: acceso, modificación y recorrido (Tema 4).
equipo = {"codigo": "SRV01", "tipo": "Servidor", "unidades": 4}
print(equipo["tipo"], len(equipo))
print(equipo.get("precio"), equipo.get("precio", 0))
print("codigo" in equipo, "SRV01" in equipo)

# Agregar, modificar y eliminar
equipo["garantia"] = True
equipo["unidades"] = 6
print(equipo)
equipo.update({"ubicacion": "Sala 2"})
print(equipo["ubicacion"])
eliminado = equipo.pop("garantia")
del equipo["ubicacion"]
print(eliminado, equipo)

# Recorrido
for clave, valor in equipo.items():
    print(f"{clave}: {valor}")
print(list(equipo.keys()))
print(list(equipo.values()))

# Contador con get()
tipos = ["Switch", "PC", "Switch", "Router", "Switch", "PC"]
cuenta = {}
for tipo in tipos:
    cuenta[tipo] = cuenta.get(tipo, 0) + 1
print(cuenta)
```

## Copia de seguridad de un inventario

La copia debería modificarse sin alterar el inventario original. Ejecuta la celda y comprueba la salida. Después sustituye `inventario.copy()` por `copy.deepcopy(inventario)` y ejecútala de nuevo.

## Lista de diccionarios

Recorre un inventario representado como una lista de diccionarios.

```python
# Inventario como lista de diccionarios (Tema 4).
inventario = [
    {"codigo": "SRV01", "tipo": "Servidor", "unidades": 4},
    {"codigo": "SW01", "tipo": "Switch", "unidades": 7},
    {"codigo": "SW02", "tipo": "Switch", "unidades": 0},
    {"codigo": "PC01", "tipo": "Portátil", "unidades": 12},
    {"codigo": "RT01", "tipo": "Router", "unidades": 3},
]

# Recorrido con acceso a las claves
for equipo in inventario:
    print(f"{equipo['codigo']:<6} {equipo['tipo']:<10} {equipo['unidades']:>3}")

# Total de unidades
total = 0
for equipo in inventario:
    total += equipo["unidades"]
print("Total de unidades:", total)

# Códigos con menos de 5 unidades
bajos = []
for equipo in inventario:
    if equipo["unidades"] < 5:
        bajos.append(equipo["codigo"])
print("Stock bajo:", bajos)

# Tipos distintos
tipos = set()
for equipo in inventario:
    tipos.add(equipo["tipo"])
print(sorted(tipos))

# Códigos agrupados por tipo
grupos = {}
for equipo in inventario:
    tipo = equipo["tipo"]
    if tipo not in grupos:
        grupos[tipo] = []
    grupos[tipo].append(equipo["codigo"])
for tipo, codigos in grupos.items():
    print(tipo, codigos)
```
