## Listas frente a arrays

Compara la media y el operador `*` en una lista y en un array.

```python
"""Listas frente a arrays de NumPy (Tema 8)."""
import numpy as np

latencias_lista = [12.4, 14.1, 18.7, 21.3]

# Media con un bucle for y con NumPy
suma = 0
for latencia in latencias_lista:
    suma = suma + latencia
print(suma / len(latencias_lista))
latencias = np.array(latencias_lista)
print(latencias.mean())

# El operador * se comporta de forma distinta
print(latencias_lista * 2)
print(latencias * 2)

# Tipo del array y tipo común de sus elementos
print(type(latencias), latencias.dtype)
mezcla = np.array([1, 2.5, 3])
print(mezcla, mezcla.dtype)
```

## Crear arrays y cambiar su forma

Crea arrays, consulta `shape`, `ndim`, `size` y `dtype`, y cambia la forma con `reshape()` y `T`.

```python
"""Crear arrays, consultar sus atributos y cambiar su forma (Tema 8)."""
import numpy as np

# Funciones para crear arrays
print(np.array([12.4, 14.1, 18.7]))
print(np.zeros(3))
print(np.ones(3, dtype=int))
print(np.arange(8, 14))
print(np.arange(0, 10, 3))
print(np.linspace(0, 1, 5))

# Atributos de un array de dos dimensiones
latencias = np.array([[12.4, 8.1, 15.2],
                      [14.1, 9.0, 16.8]])
print(latencias.shape, latencias.ndim, latencias.size)
print(latencias.dtype)

# Conversión de tipo
print(latencias.astype(int))
print(np.round(latencias).astype(int))

# Cambio de forma y traspuesta
print(latencias.reshape(3, 2))
print(latencias.reshape(-1))
print(latencias.T)
```

## Indexación y slicing

Selecciona elementos, filas y columnas con índices y slices.

```python
"""Indexación y slicing en una y en dos dimensiones (Tema 8)."""
import numpy as np

# Una dimensión
latencias = np.array([12.4, 14.1, 18.7, 21.3, 35.8, 29.4])
print(latencias[0], latencias[-1])
print(latencias[1:4])
print(latencias[::2])
print(latencias[::-1])

# Dos dimensiones, con una fila por hora y una columna por servidor
m = np.array([[12.4, 8.1, 15.2],
              [14.1, 9.0, 16.8],
              [18.7, 11.5, 22.4],
              [21.3, 13.2, 25.1]])
print(m[1, 2])
print(m[0])
print(m[:, 1])
print(m[1:, :2])

# Dos corchetes encadenados dan otro resultado
print(m[:][1])
```

## Vistas y copias

Modifica un slice y comprueba cuándo cambia el array original.

```python
"""Vistas y copias de un array (Tema 8)."""
import numpy as np

latencias = np.array([12.4, 14.1, 18.7, 21.3, 35.8, 29.4])

# Un slice es una vista que comparte la memoria con el original
vista = latencias[1:4]
vista[0] = 99.0
print(latencias)
print(vista.base is latencias)

# Con copy() los datos son independientes
copia = latencias[1:4].copy()
copia[0] = 0.0
print(latencias)
print(copia)
```

## Vectorización y broadcasting

Sustituye un bucle `for` por operaciones sobre todo el array y opera con arrays de distinta forma.

```python
"""Operaciones vectorizadas y broadcasting (Tema 8)."""
import numpy as np

# Un bucle for frente a una operación vectorizada
lista_web = [12.4, 14.1, 18.7]
lista_db = [8.1, 9.0, 11.5]
diferencias = []
for posicion in range(len(lista_web)):
    diferencias.append(round(lista_web[posicion] - lista_db[posicion], 1))
print(diferencias)

web = np.array(lista_web)
db = np.array(lista_db)
print(np.round(web - db, 1))
print(np.round(web / db, 2))

# Broadcasting con un escalar, una fila y una columna
m = np.array([[12.4, 8.1, 15.2],
              [14.1, 9.0, 16.8]])
print(m[0] / 1000)
objetivo = np.array([10.0, 8.0, 12.0])
print(m - objetivo)
retardo = np.array([[0.5], [1.0]])
print(m + retardo)
```

## Máscaras y condiciones

Filtra con máscaras booleanas, combina condiciones y usa `np.where()`.

```python
"""Máscaras booleanas, condiciones compuestas y np.where (Tema 8)."""
import numpy as np

latencias = np.array([12.4, 14.1, 18.7, 21.3, 35.8, 29.4])

# Una máscara filtra sin bucles
mascara = latencias > 20
print(mascara)
print(latencias[mascara])
print(mascara.sum())

# Condiciones compuestas, con cada condición entre paréntesis
print(latencias[(latencias > 15) & (latencias < 30)])
print(latencias[(latencias < 13) | (latencias > 30)])
print(latencias[~(latencias > 20)])

# np.where elige entre dos valores según la condición
print(np.where(latencias > 20, "alta", "normal"))
print(np.where(latencias > 20, 20.0, latencias))

# Una máscara también modifica los valores que la cumplen
latencias[latencias > 30] = 30.0
print(latencias)
```

## Agregaciones, axis y orden

Resume arrays con `sum()`, `mean()`, `std()` y `argmax()`, con el parámetro `axis`, y ordena con `np.argsort()`.

```python
"""Agregaciones, parámetro axis, orden y unión de arrays (Tema 8)."""
import numpy as np

# Agregaciones sobre un array de una dimensión
latencias = np.array([12.4, 14.1, 18.7, 21.3, 35.8, 29.4])
horas = np.arange(8, 14)
print(round(latencias.sum(), 1), latencias.mean())
print(latencias.min(), latencias.max())
print(round(latencias.std(), 2))
print(latencias.argmax(), horas[latencias.argmax()])

# El parámetro axis, con las horas en las filas y los servidores en las
# columnas
matriz = np.array([[12.4, 8.1, 15.2],
                   [14.1, 9.0, 16.8],
                   [18.7, 11.5, 22.4],
                   [21.3, 13.2, 25.1]])
print(matriz.mean(axis=0).round(1))
print(matriz.mean(axis=1).round(1))
print(matriz.max(axis=0))
print(matriz.argmax(axis=0))

# Ordenar y combinar
servidores = np.array(["web01", "db01", "app01"])
medias = np.array([16.6, 10.4, 19.9])
print(np.sort(medias)[::-1])
orden = np.argsort(medias)
print(orden, servidores[orden])
web = np.array([12.4, 14.1, 18.7])
db = np.array([8.1, 9.0, 11.5])
print(np.concatenate([web, db]))
```

Carga `Tema08/datos/latencias.csv`, calcula la media de cada servidor y guarda el resultado en `Tema08/medias.csv`.

```python
"""Cargar y guardar datos numéricos con NumPy (Tema 8)."""
import numpy as np

# Cargar el fichero, sin la fila de cabecera
datos = np.loadtxt("Tema08/datos/latencias.csv", delimiter=",",
                   skiprows=1)
print(datos.shape)
print(datos[0])

# Separar la hora de las latencias de los servidores
horas = datos[:, 0]
latencias = datos[:, 1:]
print(horas[:5])
print(latencias.shape)

# Media de cada servidor y guardado en un fichero
medias = latencias.mean(axis=0)
print(medias.round(1))
np.savetxt("Tema08/medias.csv", medias, fmt="%.1f")
```

## Valores ausentes

Carga un fichero con celdas `N/D` con `np.genfromtxt()` y filtra los `nan`.

```python
"""Valores ausentes en un fichero de datos (Tema 8)."""
import numpy as np

ruta = "Tema08/datos/latencias_ausentes.csv"

# loadtxt no admite celdas sin un número
try:
    np.loadtxt(ruta, delimiter=",", skiprows=1)
except ValueError as error:
    print("Error:", error)

# genfromtxt convierte la celda N/D en np.nan
datos = np.genfromtxt(ruta, delimiter=",", skip_header=1)
web02 = datos[:, 2]
print(web02[8:11])
print(web02.mean())
print(round(np.nanmean(web02), 1))

# Filtrar los valores ausentes con una máscara
validos = ~np.isnan(web02)
print(validos[8:11])
print(round(web02[validos].mean(), 1))
```

## Informe de latencias

El script integrador calcula medias, horas por encima de un umbral, el orden de los servidores y la hora más cargada, y guarda las medias en `Tema08/informe_medias.csv`.

```python
"""Informe de latencias por servidor y por hora (Tema 8)."""
import numpy as np

datos = np.genfromtxt("Tema08/datos/latencias_ausentes.csv",
                      delimiter=",", skip_header=1)
servidores = np.array(["web01", "web02", "db01", "app01", "dns01"])
horas = datos[:, 0]
latencias = datos[:, 1:]
umbral = 20.0

# Media de cada servidor, sin contar los valores ausentes
medias = np.nanmean(latencias, axis=0)

# Horas en que cada servidor supera el umbral. Un nan nunca lo supera
superan = (latencias > umbral).sum(axis=0)

print("Horas registradas:", len(horas))
orden = np.argsort(medias)[::-1]
for posicion in orden:
    print(servidores[posicion], round(medias[posicion], 1),
          superan[posicion])

# Hora más cargada, según la media de todos los servidores
medias_hora = np.nanmean(latencias, axis=1)
posicion_pico = medias_hora.argmax()
print("Hora más cargada:", int(horas[posicion_pico]),
      f"({medias_hora[posicion_pico]:.1f} ms)")

np.savetxt("Tema08/informe_medias.csv", medias, fmt="%.1f")
```
