## Capturar excepciones

Captura excepciones con `try` y `except`, compara el `except` de tipo concreto con el genérico y prueba los bloques `else` y `finally`.

```python
# Captura de excepciones con try, except, else y finally (Tema 6).

# Un try con un except de tipo concreto
texto = "80a"
try:
    puerto = int(texto)
    print("Puerto válido:", puerto)
except ValueError:
    print("El puerto debe ser un número")
print("El script continúa")

# Con as se guarda la excepción para mostrar su mensaje
try:
    puerto = int(texto)
except ValueError as error:
    print(error)

# Varios tipos de excepción agrupados en una tupla
for valor in ["22", None, "x"]:
    try:
        print(int(valor))
    except (ValueError, TypeError) as error:
        print(type(error).__name__)

# Un except sin tipo oculta el nombre mal escrito textoo
texto = "80"
try:
    puerto = int(textoo)
except:
    print("Puerto no válido")

# Los bloques else y finally
for texto in ["443", "https"]:
    try:
        puerto = int(texto)
    except ValueError:
        print(texto, "no es un puerto")
    else:
        print("Puerto", puerto, "correcto")
    finally:
        print("Revisado:", texto)
```

## Lanzar excepciones con raise

La función lanza una excepción con `raise` y el código que la llama la trata con `try`.

```python
# Lanzar excepciones con raise (Tema 6).
def validar_puerto(puerto):
    if not 1 <= puerto <= 65535:
        raise ValueError(f"Puerto fuera de rango: {puerto}")
    return puerto


try:
    validar_puerto(70000)
except ValueError as error:
    print("Error:", error)

print(validar_puerto(443))
```

## Leer un fichero de texto

Abre el registro con `open()` y `with`, y recórrelo con `read()`, `readline()`, `readlines()` y un bucle `for`.

```python
# Lectura de un fichero de texto con open() y with (Tema 6).

ruta = "Tema06/datos/accesos.log"

# open() devuelve un objeto fichero que hay que cerrar con close()
fichero = open(ruta, "r", encoding="utf-8")
print(type(fichero))
print(fichero.name, fichero.mode)
contenido = fichero.read()
print(contenido[:19])
fichero.close()
print(fichero.closed)

# with cierra el fichero al salir del bloque
with open(ruta, encoding="utf-8") as fichero:
    primera = fichero.readline()
    print(fichero.closed)
print(primera.strip())
print(fichero.closed)

# read() devuelve todo el contenido en una cadena
with open(ruta, encoding="utf-8") as fichero:
    contenido = fichero.read()
print(type(contenido))
print(len(contenido.splitlines()))

# readline() lee una línea y readlines() las que quedan
with open(ruta, encoding="utf-8") as fichero:
    primera = fichero.readline()
    restantes = fichero.readlines()
print(repr(primera))
print(len(restantes))

# Cada línea conserva su salto de línea, por lo que print() deja un hueco
print("--- Sin strip()")
with open(ruta, encoding="utf-8") as fichero:
    for linea in fichero:
        print(linea)

# strip() elimina el salto de línea
print("--- Con strip()")
with open(ruta, encoding="utf-8") as fichero:
    for linea in fichero:
        print(linea.strip())

# enumerate() numera las líneas y la condición omite las vacías
print("--- Con enumerate()")
with open(ruta, encoding="utf-8") as fichero:
    for numero, linea in enumerate(fichero, start=1):
        if linea.strip():
            print(numero, linea.strip())
```

## Escribir y agregar

El modo `"w"` crea o vacía el fichero y el modo `"a"` agrega al final. El fichero `Tema06/resumen.txt` es nuevo en el repositorio.

```python
# Escribir y agregar contenido a un fichero de texto (Tema 6).

ruta = "Tema06/resumen.txt"

# El modo "w" crea el fichero o lo vacía si ya existe
with open(ruta, "w", encoding="utf-8") as fichero:
    fichero.write("Resumen de peticiones\n")
    caracteres = fichero.write("10.0.0.5 4\n")
    print(caracteres)
    print("10.0.0.7 3", file=fichero)

# El modo "a" agrega al final sin borrar
with open(ruta, "a", encoding="utf-8") as fichero:
    fichero.write("10.0.0.9 3\n")

with open(ruta, encoding="utf-8") as fichero:
    print(fichero.read())

# Abrir con "w" vacía el fichero aunque no se escriba nada
with open(ruta, "w", encoding="utf-8") as fichero:
    fichero.write("")
with open(ruta, encoding="utf-8") as fichero:
    print(len(fichero.read()))
```

## Módulos estándar y functools

Importa módulos de la biblioteca estándar y combina ítems con `reduce()`.

```python
# Módulos de la biblioteca estándar y functools (Tema 6).
import math
from datetime import datetime
import statistics as estadistica
from functools import reduce

print(math.ceil(2.1))

fecha = datetime(2026, 11, 10, 9, 30)
print(fecha.strftime("%d/%m/%Y %H:%M"))

print(estadistica.mean([120, 150, 90]))

# reduce() combina los ítems de un iterable en un único valor
print(reduce(lambda total, valor: total + valor, [3, 4, 5]))
print(sum([3, 4, 5]))


def sumar_contadores(acumulado, parcial):
    for direccion, peticiones in parcial.items():
        acumulado[direccion] = (acumulado.get(direccion, 0)
                                + peticiones)
    return acumulado


# El tercer argumento de reduce() es el valor inicial del acumulado
lunes = {"10.0.0.5": 3, "10.0.0.7": 2}
martes = {"10.0.0.5": 1, "10.0.0.9": 4}
miercoles = {"10.0.0.7": 5}
semana = reduce(sumar_contadores, [lunes, martes, miercoles], {})
print(semana)
```

## Peticiones HTTP con requests

Consulta la API de Ollama y trata los errores de estado y de red. Debe estar instalado `requests` en el entorno `.venv` y el servicio de Ollama activo.

```python
# Lista los modelos de Ollama con la biblioteca requests (Tema 6).
import requests

# API de Ollama: esta máquina, puerto 11434
url = "http://127.0.0.1:11434/api/tags"
respuesta = requests.get(url, timeout=5)
print(respuesta.status_code, respuesta.ok)
datos = respuesta.json()
print(type(datos))
for modelo in datos["models"]:
    print(modelo["name"], round(modelo["size"] / 1e9, 1), "GB")
```

```python
# Códigos de estado y errores de red con requests (Tema 6).
import requests


def listar_modelos(url):
    try:
        respuesta = requests.get(url, timeout=5)
        respuesta.raise_for_status()
    except requests.HTTPError as error:
        print("Error HTTP:", error)
        return []
    except requests.RequestException as error:
        print("No se puede conectar:", type(error).__name__)
        return []
    return respuesta.json()["models"]


base = "http://127.0.0.1:11434"
print(len(listar_modelos(base + "/api/tags")))
print(listar_modelos(base + "/api/noexiste"))
```

## Informe de peticiones

La misma función del script `informe_peticiones.py`, aplicada al registro de accesos.

```python
# Informe de peticiones por dirección IP de un registro.
RUTA_PREDETERMINADA = "Tema06/datos/accesos.log"


def contar_peticiones(lineas):
    """Devuelve las peticiones de cada dirección y las descartadas."""
    totales = {}
    descartadas = 0
    for linea in lineas:
        try:
            direccion, metodo, pagina = linea.split()
        except ValueError:
            descartadas += 1
        else:
            totales[direccion] = totales.get(direccion, 0) + 1
    return totales, descartadas


with open(RUTA_PREDETERMINADA, encoding="utf-8") as fichero:
    lineas = [linea.strip() for linea in fichero if linea.strip()]
totales, descartadas = contar_peticiones(lineas)
for direccion in sorted(totales):
    print(direccion, totales[direccion])
print("Líneas descartadas:", descartadas)
```
