## Comprobación del kernel

Esta celda muestra el intérprete que ejecuta el notebook. El ejecutable
debe pertenecer al entorno virtual `.venv` del repositorio.

```python
import os
import platform
import sys

print("Notebook ejecutado correctamente")
print("Ejecutable:", sys.executable)
print("Version:", sys.version.split()[0])
print("Directorio:", os.getcwd())
print("Sistema:", platform.platform())
```

## Sangría

Python delimita los bloques con una sangría de 4 espacios.

```python
equipos = 3
if equipos > 0:
    print("Hay equipos")
else:
    print("Sin equipos")
```

## Tipado dinámico y fuerte

Una variable puede cambiar de tipo, pero Python no mezcla tipos
incompatibles. Para unir un texto y un número hay que convertirlo.

```python
inventario = 12
print(type(inventario))

inventario = "doce"
print(type(inventario))

print("Equipos: " + str(12))
```
