## Tipos, conversiones e identidad

Muestra el tipo de los valores básicos, convierte entre tipos y compara valores e identidades.

## Entrada de datos y f-strings

La celda pide tres datos y muestra una ficha. Escribe cada dato en el cuadro que aparece en la parte superior de la ventana y pulsa Intro.

```python
# Ficha de un equipo del inventario (Tema 3)
equipo = input("Equipo: ").strip().upper()
unidades = int(input("Unidades: "))
precio = float(input("Precio en euros: "))

total = unidades * precio

print("-" * 24)
print(f"Equipo:   {equipo}")
print(f"Unidades: {unidades:>8d}")
print(f"Precio:   {precio:>8,.2f} €")
print(f"Total:    {total:>8,.2f} €")
print("-" * 24)
```

## Condicionales

Ejecuta la celda varias veces con 0, 7, 10 y 15 unidades.

```python
# Estado del stock de un equipo (Tema 3)
unidades = int(input("Unidades: "))
garantia = input("¿En garantía? (s/n): ").strip().lower() == "s"

if unidades == 0:
    estado = "Sin existencias"
elif unidades < 10:
    estado = "Stock bajo"
else:
    estado = "Stock correcto"
print("Estado:", estado)

# condición compuesta con and
if unidades > 0 and garantia:
    print("Se puede prestar")
else:
    print("No se puede prestar")
```

## Bucles

Cambia los argumentos de `range()` y comprueba cómo cambia la salida.

```python
# Bucles for y while (Tema 3)
codigo = "SRV-01-A"
serie = "SRV01"

# for con range
for n in range(1, 4):
    print("Equipo", n)

# for con una cadena y enumerate
for pos, car in enumerate(serie, start=1):
    print(pos, car)

# continue: ignora los guiones
limpio = ""
for car in codigo:
    if car == "-":
        continue
    limpio += car
print(limpio)

# break: se detiene en el primer dígito
for car in codigo:
    if car.isdigit():
        print("Primer dígito:", car)
        break

# while con contador
intentos = 0
while intentos < 3:
    intentos += 1
    print("Intento", intentos)
```

## Script completo

Pide los datos de tres equipos y clasifica su stock.

```python
# Registro de equipos del almacén (Tema 3)
total = 0
for numero in range(1, 4):
    equipo = input(f"Equipo {numero}: ").strip().upper()
    unidades = int(input("Unidades: "))
    total += unidades
    # el umbral de stock bajo es 10 unidades
    if unidades == 0:
        estado = "sin existencias"
    elif unidades < 10:
        estado = "stock bajo"
    else:
        estado = "stock correcto"
    print(f"{equipo}: {unidades} unidades ({estado})")
print(f"Total de unidades: {total}")
```

## Valores límite del umbral de stock

Con 10 unidades el stock debe ser correcto. Mientras la condición use `<=`, la prueba de 10 unidades mostrará ERROR. Si pides al modelo que corrija el condicional, modifica la condición de la celda y ejecútala de nuevo.

```python
# Comprueba el umbral de stock con valores límite (Tema 3)
pruebas = [
    (0, "Sin existencias"),
    (9, "Stock bajo"),
    (10, "Stock correcto"),
    (11, "Stock correcto"),
]

for unidades, esperado in pruebas:
    if unidades == 0:
        obtenido = "Sin existencias"
    elif unidades <= 10:
        obtenido = "Stock bajo"
    else:
        obtenido = "Stock correcto"
    estado = "OK" if obtenido == esperado else "ERROR"
    print(estado, unidades, "esperado", esperado, "obtenido", obtenido)
```
